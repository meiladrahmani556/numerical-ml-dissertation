```python
# Notebook 03 – Exploratory Data Analysis (EDA)

## Purpose
The purpose of this notebook is to perform Exploratory Data Analysis (EDA) on the selected numerical dataset.

EDA is used to:
- Understand the structure of the dataset
- Identify missing values
- Analyse distributions of variables
- Identify potential relationships between features
- Detect data quality issues before modelling
```

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
sns.set(style="whitegrid")
```

```python
# Load a temporary numerical dataset (Iris dataset)
from sklearn.datasets import load_iris

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['target'] = iris.target

df.head()
```

```python
# Dataset shape
print("Dataset shape:", df.shape)

# Dataset information
df.info()
```

```python
# Check missing values
missing_values = df.isnull().sum()
missing_values[missing_values > 0]
```

```python
# Statistical summary
df.describe()
```

```python
## Target Variable Analysis

This section analyses the distribution of the target variable to understand:
- Class balance (classification problem)
- Whether resampling techniques may be required
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='target', data=df)
plt.title("Target Variable Distribution")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

```python
df.hist(figsize=(12, 8))
plt.suptitle("Feature Distributions", y=1.02)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title("Feature Correlation Matrix")
plt.show()
```

```python
## Summary of EDA Findings

- The dataset contains only numerical features
- No missing values were detected
- Feature distributions are well-behaved
- The target variable is reasonably balanced
- Several features show moderate correlation, which may influence model selection

These insights will guide preprocessing and model development in the next stage.
```
